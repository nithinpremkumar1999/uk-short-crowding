```python
import os
import numpy as np
import pandas as pd
```

```python
agg_crowding_base_df = pd.read_parquet(os.path.join("../data/processed", "agg_base_crowding.parquet"))
agg_crowding_strict_df = pd.read_parquet(os.path.join("../data/processed", "agg_strict_crowding.parquet"))
```

```python
agg_crowding_base_df.head()
```

```python
crowding_cols = ["n_holders", "n_eff", "overlap_deg"]
correlation = agg_crowding_base_df[crowding_cols].corrwith(agg_crowding_base_df["agg_short"],\
                                                            method="spearman")
print(correlation); print(correlation.median())
```

```python
crowding_cols = ["n_holders", "n_eff", "overlap_deg"]
correlation = agg_crowding_strict_df[crowding_cols].corrwith(agg_crowding_strict_df["agg_short"],\
                                                            method="spearman")
print(correlation)
```
